# Task 3

```python
import numpy as np
```

```python
# Convert DataFrame to NumPy arrays
X_train_manual = X_train.values
X_test_manual = X_test.values

y_train_manual = y_train.values
y_test_manual = y_test.values
```

```python
# Feature Scaling
mean = X_train_manual.mean(axis=0)
std = X_train_manual.std(axis=0)

X_train_manual = (X_train_manual - mean) / std
X_test_manual = (X_test_manual - mean) / std
```

```python
# Add Bias Term
X_train_manual = np.c_[np.ones((X_train_manual.shape[0], 1)), X_train_manual]
X_test_manual = np.c_[np.ones((X_test_manual.shape[0], 1)), X_test_manual]
```

```python
# Initialize Parameters
learning_rate = 0.01
iterations = 1000

m = X_train_manual.shape[0]
n = X_train_manual.shape[1]

theta = np.zeros(n)
```

```python
# Train Linear Regression using Gradient Descent

for i in range(iterations):
    predictions = np.dot(X_train_manual, theta)

    error = predictions - y_train_manual

    gradients = (1 / m) * np.dot(X_train_manual.T, error)

    theta = theta - learning_rate * gradients
```

```python
# Predict

manual_prediction = np.dot(X_test_manual, theta)
```

```python
# Evaluate Manual Model

mae = np.mean(np.abs(y_test_manual - manual_prediction))
mse = np.mean((y_test_manual - manual_prediction) ** 2)

ss_total = np.sum((y_test_manual - np.mean(y_test_manual)) ** 2)
ss_residual = np.sum((y_test_manual - manual_prediction) ** 2)

r2 = 1 - (ss_residual / ss_total)

print("Manual MAE :", mae)
print("Manual MSE :", mse)
print("Manual R2 :", r2)
```

# Task 5

## 1. Model Evaluation (RSS, RMSE, R²)

```python
from sklearn.metrics import mean_squared_error, r2_score

pred_train = model.predict(X_train)
pred_test = model.predict(X_test)

rss_test = np.sum((y_test - pred_test)**2)
rmse_test = np.sqrt(mean_squared_error(y_test, pred_test))
r2_test = r2_score(y_test, pred_test)

print("Linear Regression - Test Set")
print(f"RSS  : {rss_test:.4f}")
print(f"RMSE : {rmse_test:.4f}")
print(f"R2   : {r2_test:.4f}")
```

## 2. Check Overfitting / Underfitting

```python
train_r2 = r2_score(y_train, pred_train)
test_r2 = r2_score(y_test, pred_test)

print(f"Train R2 : {train_r2:.4f}")
print(f"Test R2  : {test_r2:.4f}")
print(f"Difference : {train_r2 - test_r2:.4f}")

if train_r2 - test_r2 > 0.1:
    print("-> Likely OVERFITTING (train much higher than test)")
elif train_r2 < 0.5 and test_r2 < 0.5:
    print("-> Likely UNDERFITTING (both scores low)")
else:
    print("-> Good fit (train and test scores close)")
```

## 3. Cross-Validation (5-Fold)

```python
from sklearn.model_selection import cross_val_score, KFold

kfold = KFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(model, X_train, y_train, cv=kfold, scoring='r2')

print("5-Fold CV R2 scores:", cv_scores)
print(f"Average R2 : {cv_scores.mean():.4f}")
print(f"Std Dev    : {cv_scores.std():.4f}  (spread — lower is more stable)")
```

## 4. Compare All Models

```python
from sklearn.ensemble import RandomForestRegressor, AdaBoostRegressor, GradientBoostingRegressor

models = {
    "Linear Regression": LinearRegression(),
    "Random Forest": RandomForestRegressor(random_state=42),
    "AdaBoost": AdaBoostRegressor(random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42)
}

results = []
for name, m in models.items():
    m.fit(X_train, y_train)
    pred = m.predict(X_test)

    rss = np.sum((y_test - pred)**2)
    rmse = np.sqrt(mean_squared_error(y_test, pred))
    r2 = r2_score(y_test, pred)

    cv = cross_val_score(m, X_train, y_train, cv=kfold, scoring='r2')

    results.append({
        "Model": name, "RSS": rss, "RMSE": rmse, "R2": r2,
        "CV Mean R2": cv.mean(), "CV Std": cv.std()
    })

results_df = pd.DataFrame(results)
results_df
```

```python
best_row = results_df.sort_values("CV Mean R2", ascending=False).iloc[0]
print("Best model based on CV R2 (score + stability):", best_row["Model"])
results_df.sort_values("CV Mean R2", ascending=False)
```

## 5. Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV

best_model_name = best_row["Model"]
print("Tuning:", best_model_name)

if best_model_name == "Random Forest":
    param_grid = {'n_estimators': [50, 100, 200], 'max_depth': [None, 5, 10, 20], 'min_samples_split': [2, 5, 10]}
    base_model = RandomForestRegressor(random_state=42)
elif best_model_name == "Gradient Boosting":
    param_grid = {'n_estimators': [50, 100, 200], 'learning_rate': [0.01, 0.05, 0.1], 'max_depth': [2, 3, 5]}
    base_model = GradientBoostingRegressor(random_state=42)
elif best_model_name == "AdaBoost":
    param_grid = {'n_estimators': [50, 100, 200], 'learning_rate': [0.01, 0.1, 1.0]}
    base_model = AdaBoostRegressor(random_state=42)
else:  # Linear Regression has no real hyperparameters to tune
    param_grid = {'fit_intercept': [True, False]}
    base_model = LinearRegression()

grid_search = GridSearchCV(base_model, param_grid, cv=5, scoring='r2', n_jobs=-1)
grid_search.fit(X_train, y_train)

print("Best Params :", grid_search.best_params_)
print("Best CV R2  :", grid_search.best_score_)

tuned_model = grid_search.best_estimator_
```

```python
tuned_pred = tuned_model.predict(X_test)

tuned_rss = np.sum((y_test - tuned_pred)**2)
tuned_rmse = np.sqrt(mean_squared_error(y_test, tuned_pred))
tuned_r2 = r2_score(y_test, tuned_pred)

print(f"Tuned {best_model_name} - Test Set")
print(f"RSS  : {tuned_rss:.4f}")
print(f"RMSE : {tuned_rmse:.4f}")
print(f"R2   : {tuned_r2:.4f}")

print(f"\nBefore tuning R2 : {best_row['R2']:.4f}")
print(f"After tuning R2  : {tuned_r2:.4f}")
print("Improved!" if tuned_r2 > best_row['R2'] else "No further improvement found")
```
